```python
import numpy as np
import os
from PIL import Image
```

```python
def FindDepthMinMaxDifferences(pairs):
    differences = []

    for depth_gt_path, rgb_path, depth_lr_path in pairs:

        # Load GT and LR depths
        gt_depth = np.array(Image.open(depth_gt_path)).astype(np.float32) / 1000.0
        lr_depth = np.array(Image.open(depth_lr_path)).astype(np.float32) / 1000.0

        # Calculate independently
        gt_min = gt_depth.min()
        gt_max = gt_depth.max()

        lr_min = lr_depth.min()
        lr_max = lr_depth.max()

        # Check if either min or max is different
        if gt_min != lr_min or gt_max != lr_max:
            differences.append({
                "gt_path": depth_gt_path,
                "lr_path": depth_lr_path,
                "gt_min": gt_min,
                "gt_max": gt_max,
                "lr_min": lr_min,
                "lr_max": lr_max
            })

    return differences
```

```python
from Utilities.ProcessingTOFDSRReal import ProcessingTOFDSRReal

train_pairs, test_pairs = ProcessingTOFDSRReal._LoadPaths()

train_differences = FindDepthMinMaxDifferences(train_pairs)
test_differences = FindDepthMinMaxDifferences(test_pairs)

print("Train differences:", len(train_differences))
print("Test differences:", len(test_differences))

for item in train_differences[:10]:
    print(item)
```

```python
diff_min = []
diff_max = []

for item in train_differences:
    diff_min.append(abs(item['gt_min'] - item['lr_min']))
    diff_max.append(abs(item['gt_max'] - item['lr_max']))

print(sum(diff_max) / len(diff_max))
print(sum(diff_min) / len(diff_min))
```
