```python
import os
import h5py
import numpy as np
from PIL import Image
from collections import defaultdict
from tqdm import tqdm

def read_sunrgbd_depth(path, max_depth=8.0):
    """Decode a SUN RGB-D depth PNG into a float32 depth map in metres."""
    d = np.array(Image.open(path)).astype(np.uint16)
    # Undo the 3-bit right rotation used by SUN RGB-D, then mask back to 16 bits.
    d = ((d >> 3) | (d << 13)) & 0xFFFF
    d = d.astype(np.float32) / 1000.0          # millimetres -> metres
    d[d > max_depth] = max_depth               # clamp far/invalid returns
    return d     

def GetRGBDepthPathPairs(base):
    u = defaultdict(dict)
    # For each file in the directory
    count = 0
    for root, dirs, files in os.walk(base):
        for file in files:
            
            filepath = os.path.join(root, file)

            if '.jpg' in filepath:
                count += 1

            res = filepath.split('/')
            # Create a dict key to uniquely identify frames
            r1 = res[-4] + '_' + res[-3]

            # Assumption is for every rgb there is a depth
            if file.lower().endswith(('.png')) and '/depth/' in filepath:
                u[r1]['depth'] = filepath
            elif file.lower().endswith(('.jpg')) and '/image/' in filepath:
                u[r1]['rgb'] = filepath

    # Convert the default dict to a list
    pairs = []
    # Here implicitly it will be checked if every pair has a valid depth and rgb path, else there should be an error
    for k, v in u.items():
        pairs.append({
            'rgb': v['rgb'],
            'depth': v['depth']
        })

    return pairs, count


base = './SUN_Inp/'

test_pairs, count = GetRGBDepthPathPairs(base)
```

```python
# Load and convert the depths into nump arrays
depths = []
for i in test_pairs:
    # Load the depth png and convert to numpy
    depths.append(read_sunrgbd_depth(i['depth']))
```
